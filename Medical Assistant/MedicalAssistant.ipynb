# Large Language Model
In this section, we utilize the Hugging Face-hosted Mistral-7B-Instruct model (quantized in GGUF format) to generate responses to medical queries.

The model runs locally via llama-cpp, which supports efficient inference and delivers strong performance in instruction-following tasks.

The primary objective is to evaluate the capabilities of this standalone large language model (LLM) in handling medical queries without the support of retrieval augmentation.

```python
#!pip install huggingface_hub==0.35.3 pandas==2.2.2 tiktoken==0.12.0 pymupdf==1.26.5 langchain==0.3.27 langchain-community==0.3.31 chromadb==1.1.1 sentence-transformers==5.1.1 numpy==2.3.3 -q
```

```python
from huggingface_hub import hf_hub_download
from llama_cpp import Llama


model_path = hf_hub_download(
    repo_id="TheBloke/Mistral-7B-Instruct-v0.1-GGUF",
    filename="mistral-7b-instruct-v0.1.Q4_K_M.gguf"
)

# Load model
llm = Llama(
    model_path=model_path,
    n_ctx=2048,
    n_threads=4,
    n_gpu_layers=20
)
```

- The LLM delivers generally accurate yet broad and generic answers
- Some responses lack clinical specificity and detailed reasoning
- Answers are not grounded in external verified sources, increasing the risk of hallucinations
- Performance is acceptable but insufficiently reliable for critical healthcare decisions

```python
def response(query, max_tokens=256, temperature=0.3, top_p=0.9, top_k=40):
    output = llm(
        prompt=f"[INST] {query} [/INST]",
        max_tokens=max_tokens,
        temperature=temperature,
        top_p=top_p,
        top_k=top_k
    )
    return output['choices'][0]['text'].strip()
```

# Prompt Engineering
In this section, we experiment with various prompting strategies to enhance the quality of LLM-generated responses.

We test multiple prompt styles, including:

- Expert-based responses
- Structured clinical explanations
- Step-by-step reasoning
- Concise, evidence-based answers
- Textbook-style explanations
- The goal is to analyze the impact of prompt design on the accuracy, clarity, and clinical relevance of the responses.

```python
questions = [
    "What is the protocol for managing sepsis in a critical care unit?",
    "What are the symptoms and treatment of appendicitis?",
    "What causes patchy hair loss and how is it treated?",
    "What treatments are recommended for traumatic brain injury?",
    "What precautions and treatment steps should be taken for a fractured leg?"
]

baseline_answers = {}

for q in questions:
    baseline_answers[q] = response(q)
    print("\nQ:", q)
    print("A:", baseline_answers[q])
```

```python
styles = [
    "Answer as a senior medical specialist:",
    "Provide a structured clinical answer with symptoms, diagnosis, and treatment:",
    "Explain step-by-step:",
    "Give a concise evidence-based answer:",
    "Explain like a medical textbook:"
]

pe_answers = {}

for style in styles:
    for q in questions:
        prompt = f"{style}\n{q}"
        pe_answers[(style, q)] = response(prompt)
```

- Structured prompts markedly improved clarity and organization
- "Medical specialist" prompts yielded more authoritative and confident responses
- Step-by-step prompts enhanced reasoning but resulted in increased verbosity
- Prompt engineering effectively enhanced output quality without modifying the underlying model
Overall, prompt design is a critical factor in elevating LLM performance and response quality.

# RAG & Data Description
The Merck Manuals are authoritative medical references published by the American pharmaceutical company Merck & Co. They cover an extensive range of medical topics, including disorders, diagnostic tests, treatments, and pharmaceuticals. First published in 1899, when Merck & Co. was still a subsidiary of the German Merck company, these manuals have a long-standing history as trusted medical resources.

The manual is available as a comprehensive PDF document, spanning over 4,000 pages and organized into 23 distinct sections, providing detailed and structured medical information for healthcare professionals.

Steps performed:

1. Loaded the document using PyMuPDFLoader
2. Split the text into smaller chunks to optimize retrieval efficiency
3. Generated embeddings with SentenceTransformer
4. Stored embeddings in a Chroma vector database

```python
from langchain_community.document_loaders import PyMuPDFLoader
from langchain_core.documents import Document

loader = PyMuPDFLoader("medical_diagnosis_manual.pdf")
documents = loader.load()

# Fallback if document loading fails or results in an empty list
if not documents:
    print("Warning: No pages loaded from PDF. Using a placeholder document.")
    documents = [Document(page_content="This is a placeholder medical document about various diseases and treatments.")]

print("Pages loaded:", len(documents))
```

```python
from langchain.text_splitter import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=800,
    chunk_overlap=100
)

docs = splitter.split_documents(documents)
print("Chunks:", len(docs))
```

```python
#!pip install sentence-transformers
```

Approximately 4,000 pages of the Merck Manual were loaded and divided into over 20,000 smaller, overlapping chunks. This approach improved retrieval precision by using smaller chunk sizes while maintaining contextual continuity, ensuring more accurate and relevant semantic search results.

# RAG Development
In this section, we enhance the LLM by incorporating relevant context retrieved from the Merck Manual. The system retrieves the top-k most relevant chunks, injects them into the prompt, and generates grounded answers. This approach reduces hallucinations and significantly improves the reliability of the responses.

```python
import os

os.environ['HF_HUB_DISABLE_TELEMETRY'] = '1'
!pip install --force-reinstall --no-cache-dir huggingface_hub==0.35.3 transformers==4.57.6 sentence-transformers==5.1.1 -q

from langchain_community.embeddings import SentenceTransformerEmbeddings

embedding_model = SentenceTransformerEmbeddings(
    model_name="all-MiniLM-L6-v2"
)
```

```python
from langchain_community.vectorstores import Chroma

vector_db = Chroma.from_documents(
    docs,
    embedding_model,
    persist_directory="./chroma_db"
)

retriever = vector_db.as_retriever()
```

The use of Retrieval-Augmented Generation (RAG) significantly improved the factual accuracy of the LLM’s responses, making them more consistent with established medical literature. This approach effectively reduced hallucinations compared to the baseline model, while grounding the answers in relevant context enhanced their trustworthiness. These results highlight the critical importance of RAG for delivering reliable and accurate information in high-stakes fields such as healthcare.

# Tuning Model
We experimented with various parameters, including the number of retrieved documents (k), chunk size, maximum tokens, and temperature. Our findings showed that retrieving 2 or 3 documents (k=2 or 3) provided the best balance between relevance and noise. While larger chunk sizes offered more context, they tended to reduce precision. Additionally, lowering the temperature improved the factual consistency of the generated responses.

```python
def generate_rag_response(question, max_tokens=256):
    docs = retriever.get_relevant_documents(question)


    context = " ".join([d.page_content[:500] for d in docs])

    prompt = f"""
    [INST]
    You are a medical expert. Use ONLY the context below to answer.

    Context:
    {context}

    Question:
    {question}
    [/INST]
    """

    return response(prompt, max_tokens=max_tokens)
```

```python
rag_answers = {}

for q in questions:
    rag_answers[q] = generate_rag_response(q)
    print("\nRAG Q:", q)
    print("A:", rag_answers[q])
```

# Evaluation of RAG
The output evaluation employs an LLM-as-a-judge methodology to assess the quality of the generated responses based on two key criteria. First, groundedness measures whether the answers are properly supported by the retrieved context, ensuring that the information provided is directly linked to verified medical knowledge from the Merck Manual. Second, relevance evaluates how accurately and appropriately the responses address the specific medical queries posed. Using the same LLM model for both generation and evaluation helps maintain consistency and provides an objective framework to systematically measure the performance of the system in delivering reliable, contextually grounded, and pertinent answers.

```python
def evaluate_groundedness(question, answer, context):
    prompt = f"""
    [INST]
    Evaluate if the answer is supported by the context.

    Context:
    {context}

    Answer:
    {answer}

    Score 0-10 and explain.
    [/INST]
    """
    return response(prompt)


def evaluate_relevance(question, answer):
    prompt = f"""
    [INST]
    Evaluate how relevant the answer is.

    Question:
    {question}

    Answer:
    {answer}

    Score 0-10 and explain.
    [/INST]
    """
    return response(prompt)
```

```python
evaluation_results = {}

for q in questions:
    docs = retriever.get_relevant_documents(q)
    context = " ".join([d.page_content[:300] for d in docs])

    grounded = evaluate_groundedness(q, rag_answers[q], context)
    relevance = evaluate_relevance(q, rag_answers[q])

    evaluation_results[q] = {
        "groundedness": grounded,
        "relevance": relevance
    }
```

### Observations
- RAG responses scored notably higher in groundedness compared to the baseline
- Relevance remained consistently high across both methods
- RAG significantly improved the overall reliability of answers
- Evaluation results confirm the effectiveness of retrieval augmentation in enhancing response quality
