## **Problem Definition**

American companies are depending more and more on foreign talent to fill labor shortages in a variety of areas. The Office of Foreign Labor Certification (OFLC) is in charge of reviewing an increasing number of visa applications annually to ascertain if firms may engage foreign workers without negatively affecting the pay or working conditions of American workers. It takes more time and resources to carefully evaluate every case as the number of applications keeps growing.


The goal of this project is to create a machine learning-based classification model that forecasts whether a visa application will be approved or rejected based on employer and applicant attributes like employment region, education level, job experience, prevailing wage, and company size. This methodology seeks to help decision-makers prioritize applications with a better chance of acceptance and indicate situations that might need more examination by identifying the critical elements that affect visa approval decisions.

From a business standpoint, the solution may shorten processing times, increase operational effectiveness, and offer data-driven insights into the profiles most likely to be certified for a visa. In order to ensure that the model promotes fair and informed decision-making rather than taking the place of human judgment, the final model is assessed using a variety of performance measures, with a focus on recall and overall prediction dependability.

# **Data Pre-Processing**
To get the dataset ready for efficient modeling, data pre-processing was done. By resolving missing values, outliers, and feature changes as needed, this stage guarantees that the data is clean, consistent, and appropriate for machine learning algorithms.

```python
# Copy df for preprocessing stage
df_pre = df.copy()

# Analyze basic info
df_pre.shape, df_pre.info()
```

#### **Missing Values Handing**

```python
# Check for missing vals
df_pre.isnull().sum()
```

No missing values were found and therefore, no missing values require handling.

#### **Outlier Detection and Handling**

```python
numerical_cols = ["prevailing_wage", "no_of_employees", "yr_of_estab"]

for col in numerical_cols:
    plt.figure(figsize=(5, 2))
    sns.boxplot(x=df_pre[col])
    plt.title(f"Boxplot of {col}")
    plt.show()
```

```python
# Log transform prevailing wage due to heavy right skew
df_pre["prevailing_wage"] = np.log1p(df_pre["prevailing_wage"])
df_pre["no_of_employees"] = np.log1p(df_pre["no_of_employees"])
```

Because real-world economic data typically has right-skewed distributions, box plots show the existence of outliers in prevailing pay, number of employees, and year of founding. These outliers were kept since they don't indicate data problems but rather legitimate situations like high-paying positions and big, established companies. Skewed variables were treated with robust tree-based models that are less susceptible to extreme values and log transformation in order to lessen their impact.

## **Feature Engineering**

```python
# Convert target variable to binary
df_pre["case_status"] = df_pre["case_status"].map({"Denied": 0, "Certified": 1})
# One-hot encode categorical variables
df_pre = pd.get_dummies(df_pre, drop_first=True)
```

The target variable was transformed into a binary format appropriate for categorization as part of feature engineering. To enable machine learning models to read non-numeric data without enforcing ordinal connections, categorical variables were converted via one-hot encoding.

## **Preparation of Data for Modeling**

```python
X = df_pre.drop("case_status", axis=1)
y = df_pre["case_status"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.2, 
    stratify=y, 
    random_state=42
)
```

An 80–20 split was used to divide the dataset into training and testing sets. Because of the observed class imbalance in visa certification outcomes, stratified sampling was used to maintain the original class distribution.

# **Model Building**


## **Original Data**
Using the original (unbalanced) dataset, many classification models were constructed in this part to determine baseline performance. Because they can handle heterogeneous data types and nonlinear relationships, tree-based and ensemble models were chosen. Accuracy, precision, recall, F1-score, and ROC-AUC were used to assess the model's performance.

```python
## evaluation of model
def evaluate_model(model, X_train, y_train, X_test, y_test):
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1]
    
    return {
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred),
        "Recall": recall_score(y_test, y_pred),
        "F1": f1_score(y_test, y_pred),
        "ROC-AUC": roc_auc_score(y_test, y_prob)
    }
## defining the model(s)
models = {
    "Decision Tree": DecisionTreeClassifier(
        max_depth=10, random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=50,
        max_depth=12,
        n_jobs=-1,
        random_state=42
    ),

    "Bagging Classifier": BaggingClassifier(
        n_estimators=30,
        n_jobs=-1,
        random_state=42
    ),

    "AdaBoost": AdaBoostClassifier(
        n_estimators=30,
        random_state=42
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        n_estimators=50,
        max_depth=3,
        subsample=0.8,
        learning_rate=0.1,
        random_state=42
    )
}
## retrain/evaluate again
results_original = {}

for name, model in models.items():
    results_original[name] = evaluate_model(
        model, X_train, y_train, X_test, y_test
    )

results_original_df = pd.DataFrame(results_original).T
results_original_df
```

In the majority of assessment measures, ensemble models like Random Forest, Bagging, and Boosting perform better than a single Decision Tree. Due to class imbalance, recall for the minority class is still comparatively poor, suggesting that the model's capacity to accurately identify cases of rejected visas is constrained by the original dataset. This is why resampling techniques are used in the following sections.
